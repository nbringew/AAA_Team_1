## Load Taxi Data

The dataset `ajtu-isnz` ("Taxi Trips 2024–") contains trips starting from 2024-01-01. We load three time periods into separate CSVs: **2024**, **2025**, and **2026 (Jan–May)**. Already fully downloaded files are skipped automatically — so 2025 doesn't need to be re-downloaded.

**Why does 2026 end on 05-31?** The portal has a reporting lag of several weeks: as of early July 2026, June only contains 31 rows (instead of ~600,000). Incomplete months would look like a demand crash in the demand prediction and are therefore hard-excluded.


```python
import os
import time
import pandas as pd
from sodapy import Socrata
from datetime import date, timedelta

APP_TOKEN   = None
DATASET_ID  = "ajtu-isnz"   # "Taxi Trips (2024–)" – contains data from 2024-01-01

# pickup_centroid_location / dropoff_centroid_location are omitted –
# they duplicate the separate latitude/longitude columns as a Point object.
COLUMNS = [
    "trip_id",
    "taxi_id",
    "trip_start_timestamp",
    "trip_end_timestamp",
    "trip_seconds",
    "trip_miles",
    "pickup_census_tract",
    "dropoff_census_tract",
    "pickup_community_area",
    "dropoff_community_area",
    "fare",
    "tips",
    "tolls",
    "extras",
    "trip_total",
    "payment_type",
    "company",
    "pickup_centroid_latitude",
    "pickup_centroid_longitude",
    "dropoff_centroid_latitude",
    "dropoff_centroid_longitude",
]

CHUNK_DAYS  = 7      # weekly chunks – small enough for a fast server response
BATCH_SIZE  = 50_000
MAX_RETRIES = 8

# Download ranges: one CSV + one checkpoint per period.
# 2026 ends on 05-31 – see the reporting-lag note in the markdown above.
DOWNLOAD_RANGES = [
    (date(2024, 1, 1), date(2024, 12, 31), "../../data/chicago_taxi_2024.csv",    "../../data/download_checkpoint_2024.txt"),
    (date(2025, 1, 1), date(2025, 12, 31), "../../data/chicago_taxi_2025.csv",    "../../data/download_checkpoint_2025.txt"),
    (date(2026, 1, 1), date(2026, 5, 31),  "../../data/chicago_taxi_2026_h1.csv", "../../data/download_checkpoint_2026.txt"),
]


def load_checkpoint_date(checkpoint_file, start_date):
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file) as f:
            d = date.fromisoformat(f.read().strip())
        print(f"  Checkpoint found: starting from {d}")
        return d
    return start_date


def save_checkpoint_date(checkpoint_file, d):
    with open(checkpoint_file, "w") as f:
        f.write(d.isoformat())


def append_to_csv(df, filepath):
    write_header = not os.path.exists(filepath)
    df.to_csv(filepath, mode="a", index=False, header=write_header)


def check_columns_match(filepath, expected_columns):
    """Checks whether the columns of the existing CSV match COLUMNS."""
    with open(filepath, "r") as f:
        header = f.readline().strip().split(",")
    return header == expected_columns


def cast_types(df):
    df["trip_start_timestamp"] = pd.to_datetime(df["trip_start_timestamp"], errors="coerce")
    df["trip_end_timestamp"]   = pd.to_datetime(df["trip_end_timestamp"],   errors="coerce")
    for col in ["trip_seconds", "trip_miles", "fare", "tips", "tolls", "extras", "trip_total",
                "pickup_centroid_latitude",  "pickup_centroid_longitude",
                "dropoff_centroid_latitude", "dropoff_centroid_longitude",
                "pickup_community_area",     "dropoff_community_area"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")
    return df


def download_taxi_range(start_date, end_date, output_file, checkpoint_file):
    # ── Column consistency check ──────────────────────────────────────────────────
    # No checkpoint can mean two things: (a) never downloaded yet, or
    # (b) the last run finished completely (the checkpoint is removed at the end).
    # Only in case (a) with the wrong columns do we want to delete and restart. In
    # case (b) with matching columns we do NOT delete automatically – otherwise an
    # accidental re-run of the cell would destroy an already-complete CSV and
    # trigger a full, time-consuming re-download.
    if os.path.exists(output_file) and not os.path.exists(checkpoint_file):
        if not check_columns_match(output_file, COLUMNS):
            print(f"  WARNING: existing CSV has different columns than COLUMNS.")
            print(f"  Deleting old file and restarting download.")
            os.remove(output_file)
        else:
            print(f"  Existing CSV with matching columns found, no open checkpoint.")
            print(f"  → Presumably already fully downloaded. Skipping download.")
            print(f"  (To force a re-download, please delete {output_file} manually.)")
            return

    print("Connecting to Chicago Data Portal...")
    client = Socrata("data.cityofchicago.org", APP_TOKEN, timeout=180)

    current = load_checkpoint_date(checkpoint_file, start_date)
    select_clause = ", ".join(COLUMNS)
    total_rows = 0

    while current <= end_date:
        chunk_end = min(current + timedelta(days=CHUNK_DAYS), end_date + timedelta(days=1))
        where = (
            f"trip_start_timestamp >= '{ current.isoformat() }T00:00:00'"
            f" AND trip_start_timestamp < '{ chunk_end.isoformat() }T00:00:00'"
        )

        offset = 0
        while True:
            # retry loop
            results = None
            for attempt in range(1, MAX_RETRIES + 1):
                try:
                    results = client.get(
                        DATASET_ID,
                        select=select_clause,
                        where=where,
                        limit=BATCH_SIZE,
                        offset=offset,
                    )
                    break
                except Exception as e:
                    wait = 3 * (2 ** (attempt - 1))
                    print(f"  ✗ Attempt {attempt}/{MAX_RETRIES}: {e}")
                    if attempt < MAX_RETRIES:
                        print(f"    Waiting {wait}s...")
                        time.sleep(wait)
                    else:
                        print(f"  Aborting. Checkpoint at {current}.")
                        save_checkpoint_date(checkpoint_file, current)
                        return

            if not results:
                break  # no results in this time window

            chunk = cast_types(pd.DataFrame.from_records(results))
            chunk = chunk[COLUMNS]
            append_to_csv(chunk, output_file)

            offset     += len(results)
            total_rows += len(results)

            size_mb = os.path.getsize(output_file) / 1e6
            print(f"  ✓ {current} – {chunk_end}  |  +{len(results):,} rows  |  total {total_rows:,}  |  {size_mb:.0f} MB")

            if len(results) < BATCH_SIZE:
                break  # last batch of this time window

        save_checkpoint_date(checkpoint_file, chunk_end)  # start next week
        current = chunk_end
        time.sleep(0.3)

    if os.path.exists(output_file):
        final_size = os.path.getsize(output_file) / 1e6
        print(f"── Done ────────────────────────────────────────────")
        print(f"  File  : {os.path.abspath(output_file)}")
        print(f"  Size  : {final_size:.1f} MB")
        print(f"  Rows  : {total_rows:,}")
    if os.path.exists(checkpoint_file):
        os.remove(checkpoint_file)


for rng_start, rng_end, out_csv, ckpt in DOWNLOAD_RANGES:
    print(f"\n══ {rng_start} to {rng_end} → {out_csv} ══")
    download_taxi_range(rng_start, rng_end, out_csv, ckpt)
```

## Load weather data


```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude":  41.8781,   # Chicago
    "longitude": -87.6298,
    "start_date": "2024-01-01",
    "end_date":   "2026-05-31",   # same data cutoff as the taxi data (reporting lag from June 2026 onward)
    "hourly": [
        "temperature_2m",
        "apparent_temperature",
        "precipitation",
        "snowfall",
        "wind_speed_10m",
        "cloud_cover",
        "weather_code",
    ],
    "timezone": "America/Chicago",  # local time, matches the taxi data
    "wind_speed_unit": "mph",
}

response = requests.get(url, params=params)
data     = response.json()

# WMO Weather Code → human-readable description (Open-Meteo uses WMO standard)
WMO_CODES = {
    0:  "Clear sky",
    1:  "Mainly clear",
    2:  "Partly cloudy",
    3:  "Overcast",
    45: "Fog",
    48: "Rime fog",
    51: "Light drizzle",
    53: "Moderate drizzle",
    55: "Dense drizzle",
    56: "Light freezing drizzle",
    57: "Heavy freezing drizzle",
    61: "Slight rain",
    63: "Moderate rain",
    65: "Heavy rain",
    66: "Light freezing rain",
    67: "Heavy freezing rain",
    71: "Slight snowfall",
    73: "Moderate snowfall",
    75: "Heavy snowfall",
    77: "Snow grains",
    80: "Slight rain showers",
    81: "Moderate rain showers",
    82: "Violent rain showers",
    85: "Slight snow showers",
    86: "Heavy snow showers",
    95: "Thunderstorm",
    96: "Thunderstorm with slight hail",
    99: "Thunderstorm with heavy hail",
}

# Convert to DataFrame
df_weather = pd.DataFrame(data["hourly"])
df_weather["time"] = pd.to_datetime(df_weather["time"])
df_weather = df_weather.rename(columns={"time": "datetime"})

df_weather["weather_description"] = df_weather["weather_code"].map(WMO_CODES).fillna("Unknown")

print(f"Rows: {len(df_weather)}")   # expected: (366 + 365 + 151) days × 24h = 21,168
print(df_weather[["datetime", "weather_code", "weather_description"]].head(10))

df_weather.to_csv("../../data/chicago_weather_2024_2026_hourly.csv", index=False)
print("Saved as ../../data/chicago_weather_2024_2026_hourly.csv")
```

## Data Cleaning – Taxi Data

Before filtering or transforming anything, we first look at *what* is actually in the raw dataset: how complete are the individual columns? Which values occur — and which are missing? This analysis motivates each of the following cleaning steps.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

CSV_PATHS = [
    '../../data/chicago_taxi_2024.csv',
    '../../data/chicago_taxi_2025.csv',
    '../../data/chicago_taxi_2026_h1.csv',
]
OUT_PARQUET = '../../data/chicago_taxi_2024_2026_clean.parquet'

# Data cutoff: the portal contains data from 2024-01-01 onward; from June 2026
# the trips are still incomplete (reporting lag) → hard end date 2026-06-01 (excl.).
DATA_START    = pd.Timestamp('2024-01-01')
DATA_END_EXCL = pd.Timestamp('2026-06-01')

# taxi_id is a 128-character hex hash, trip_id a 40-character hash – across the
# ~17 million rows of all three years, that's several GB with pandas' default
# (object) dtype for just these two columns. taxi_id repeats thousands of times
# per taxi → category saves the most here. Money/distance columns don't need
# float64 precision, float32 is enough and halves their memory footprint.
DTYPES = {
    'taxi_id':                    'category',
    'payment_type':                'category',
    'company':                     'category',
    'trip_seconds':                'float32',
    'trip_miles':                  'float32',
    'pickup_census_tract':         'float32',
    'dropoff_census_tract':        'float32',
    'pickup_community_area':       'float32',
    'dropoff_community_area':      'float32',
    'fare':                        'float32',
    'tips':                        'float32',
    'tolls':                       'float32',
    'extras':                      'float32',
    'trip_total':                  'float32',
    'pickup_centroid_latitude':    'float32',
    'pickup_centroid_longitude':   'float32',
    'dropoff_centroid_latitude':   'float32',
    'dropoff_centroid_longitude':  'float32',
}

# ── Loading ──────────────────────────────────────────────────────────────────────────────────────────────
# Each file is loaded individually, immediately slimmed down via the dtype spec,
# and filtered to the date range before the next one is added. This way, all
# three raw files never have to sit uncompressed in memory at the same time –
# with limited RAM that would otherwise risk swapping/hanging.
# trip_start/end_timestamp are deliberately NOT parsed via read_csv(parse_dates=...):
# a few rows contain only a date without a time (e.g. "2026-03-26"), which makes
# the read_csv parser fall back to object dtype for the whole column.
# pd.to_datetime(..., errors='coerce') afterward handles this format mix reliably.
parts = []
for p in CSV_PATHS:
    part = pd.read_csv(p, dtype=DTYPES, low_memory=False)
    part['trip_start_timestamp'] = pd.to_datetime(part['trip_start_timestamp'], errors='coerce')
    part['trip_end_timestamp']   = pd.to_datetime(part['trip_end_timestamp'],   errors='coerce')
    part = part[(part['trip_start_timestamp'] >= DATA_START) &
                (part['trip_start_timestamp'] <  DATA_END_EXCL)]
    parts.append(part)

df = pd.concat(parts, ignore_index=True)
del parts

# Remember the raw dataset size – needed below for the dynamic summary, so the
# numbers can never again drift from what the pipeline actually does.
raw_rows, raw_cols = df.shape
cleaning_log = []

print(f"Raw dataset: {len(df):,} rows × {df.shape[1]} columns "
      f"({df['trip_start_timestamp'].min():%Y-%m-%d} to {df['trip_start_timestamp'].max():%Y-%m-%d})\n")

# ── Null values per column ────────────────────────────────────────────────────
null_counts = df.isnull().sum()
null_pcts   = (df.isnull().mean() * 100).round(1)

overview = pd.DataFrame({
    'Null Count':  null_counts,
    'Share (%)':   null_pcts,
    'Filled (%)':  (100 - null_pcts),
}).sort_values('Share (%)', ascending=False)

print(overview.to_string())

# Visualization
fig, ax = plt.subplots(figsize=(10, 5))
colors = ['#d9534f' if p > 30 else '#f0ad4e' if p > 5 else '#5cb85c' for p in overview['Share (%)']]
bars = ax.barh(overview.index, overview['Share (%)'], color=colors)
ax.set_xlabel('Share of missing values (%)')
ax.set_title('Column completeness in the raw dataset', fontweight='bold')
ax.axvline(0, color='black', linewidth=0.8)
for bar, val in zip(bars, overview['Share (%)']):
    if val > 0:
        ax.text(val + 0.3, bar.get_y() + bar.get_height()/2, f'{val:.1f}%', va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

### Completeness Check: Trips per Month

The portal populates the data with a reporting lag of several weeks — the most recent months can therefore be **incomplete** without any error message. An incomplete month looks like a genuine demand crash in the demand prediction and would distort training and evaluation.

The check below compares each month's trip count to the median across all months. Months clearly below the median (red) should fall outside `DATA_END_EXCL`, confirming the end-date cutoff.


```python
monthly_counts = df.groupby(df['trip_start_timestamp'].dt.to_period('M')).size()
median_month   = monthly_counts.median()
suspicious     = monthly_counts[monthly_counts < 0.6 * median_month]

fig, ax = plt.subplots(figsize=(14, 4))
colors = ['#d9534f' if v < 0.6 * median_month else 'steelblue' for v in monthly_counts.values]
ax.bar(monthly_counts.index.astype(str), monthly_counts.values, color=colors)
ax.axhline(median_month, color='black', linestyle='--', linewidth=1.2,
           label=f'Median = {median_month:,.0f} trips/month')
ax.set_ylabel('Trips per month')
ax.set_title('Completeness check: trips per month (red = < 60% of median → likely incomplete)',
             fontweight='bold')
ax.legend()
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

if len(suspicious) > 0:
    print("WARNING – the following months are clearly below the median (incomplete?):")
    for period, count in suspicious.items():
        print(f"  {period}: {count:,} trips ({count/median_month*100:.0f}% of median)")
    print("\n→ If these are the most recent months: move DATA_END_EXCL forward accordingly.")
else:
    print("All months are within the expected range – no indication of incomplete months.")
```

### Step 0: Duplicates & Missing IDs (before any further filtering)

Before filtering by community area, missing trip values, etc., we check for duplicates directly at the raw-data level. If we deduplicated only later (after other filters), a duplicate pair could get "torn apart" by an intermediate filter — one row stays, the other is dropped, and the duplicate check would then falsely find no more duplicates.

Rows with a missing `trip_id` are split off separately before deduplication: `drop_duplicates` treats multiple `NaN` values as "equal" and would otherwise incorrectly merge several genuine but ID-less trips into a single one.


```python
n_null_id = df['trip_id'].isna().sum()
print(f"trip_id – missing (before dedup): {n_null_id:,}  ({n_null_id/len(df)*100:.3f}%)")

before = len(df)

# Rows without a trip_id can't be checked for duplicates -> handle them
# separately, so drop_duplicates doesn't falsely treat them as duplicates of
# each other.
df_missing_id = df[df['trip_id'].isna()]
df = df[df['trip_id'].notna()].copy()

n_dupes = df.duplicated(subset='trip_id').sum()
print(f"trip_id – duplicates (among the {len(df):,} rows with an ID): {n_dupes:,}  ({n_dupes/len(df)*100:.3f}%)")

if n_dupes > 0:
    df = df.drop_duplicates(subset='trip_id', keep='first').copy()
    print(f"  → {n_dupes:,} duplicates removed, first entry kept.")
else:
    print("  → No duplicates found.")

# Rows without a trip_id remain in the dataset (deduplication isn't possible,
# but they may still be valid, distinct trips).
df = pd.concat([df, df_missing_id], ignore_index=True)

step0_removed = before - len(df)
print(f"\nRemoved: {step0_removed:,} rows → {len(df):,} remaining")
print(f"Note: {len(df_missing_id):,} row(s) without trip_id are kept (cannot be deduplicated).")

cleaning_log.append((
    "0 - Duplicates & missing IDs",
    step0_removed,
    f"Removed duplicate trip_id (kept first entry); {len(df_missing_id):,} row(s) without trip_id are kept"
))
```

### Note: Census Tracts – Completeness & Usability

The overview shows that `pickup_census_tract` and `dropoff_census_tract` are missing for around 55% of trips. This is not a random error — Chicago intentionally masks census tracts for privacy reasons when too few trips fall on a tract. Community areas offer lower geographic resolution, but with near-complete coverage.

The columns remain in the dataset; however, analyses that require full geographic coverage should rely on community areas instead.


```python
coverage_cols = ['pickup_census_tract', 'dropoff_census_tract',
                 'pickup_community_area', 'dropoff_community_area']

print("Coverage comparison: census tract vs. community area:\n")
print(f"{'Column':<30} {'Present':>12} {'Missing':>12} {'Coverage':>12}")
print("-" * 68)
for col in coverage_cols:
    n_ok  = df[col].notna().sum()
    n_mis = df[col].isna().sum()
    pct   = n_ok / len(df) * 100
    print(f"{col:<30} {n_ok:>12,} {n_mis:>12,} {pct:>11.1f}%")

# ── Additional check: do out-of-Chicago trips also lack census tracts? ───────
no_ca_mask  = df['pickup_community_area'].isna() | df['dropoff_community_area'].isna()
no_ca_check = df[no_ca_mask]

print(f"\nAdditional check – trips without a community area ({len(no_ca_check):,} rows):")
for ct_col in ['pickup_census_tract', 'dropoff_census_tract']:
    n_mis = no_ca_check[ct_col].isna().sum()
    print(f"  {ct_col:<30}: {n_mis:,} missing  ({n_mis/len(no_ca_check)*100:.1f}%)")

both_ct_missing = (no_ca_check['pickup_census_tract'].isna() & no_ca_check['dropoff_census_tract'].isna()).sum()
print(f"  Both census tracts missing:          {both_ct_missing:,}  ({both_ct_missing/len(no_ca_check)*100:.1f}%)")
print(f"\n→ Trips outside Chicago have neither a community area nor census tracts.")
print(f"  So census tracts aren't a substitute for the missing geo-information either.")
print(f"\n→ Census tract columns remain in the dataset (can be used depending on the analysis),")
print(f"  but community areas are recommended for full geographic coverage.")
```

### Step 2: Trips Within Chicago Only

Trips whose pickup or dropoff lies outside Chicago have no `community_area`. What's behind these rows — are coordinates missing there too, and with them the possibility of spatial analysis?


```python
# Trips WITHOUT a complete community area (= potentially outside Chicago)
no_ca = df[df['pickup_community_area'].isna() | df['dropoff_community_area'].isna()].copy()
in_ca = df[df['pickup_community_area'].notna() & df['dropoff_community_area'].notna()].copy()

print(f"Trips WITHOUT a complete community area: {len(no_ca):,}  ({len(no_ca)/len(df)*100:.1f}%)")
print(f"Trips WITH    a complete community area: {len(in_ca):,}  ({len(in_ca)/len(df)*100:.1f}%)")

# Coordinate coverage in both groups
coord_cols = [
    'pickup_centroid_latitude', 'pickup_centroid_longitude',
    'dropoff_centroid_latitude', 'dropoff_centroid_longitude'
]

missing_no_ca = no_ca[coord_cols].isna().any(axis=1).sum()
missing_in_ca = in_ca[coord_cols].isna().any(axis=1).sum()

print(f"\nOf the {len(no_ca):,} trips without a community area:")
print(f"  → {missing_no_ca:,} ({missing_no_ca/len(no_ca)*100:.1f}%) have at least one missing coordinate")

print(f"\nOf the {len(in_ca):,} trips with a complete community area:")
print(f"  → {missing_in_ca:,} ({missing_in_ca/len(in_ca)*100:.1f}%) have at least one missing coordinate")

print("\n→ Decision:")
print("  Trips without a complete community area are NOT excluded.")
print("  They remain in the dataset, since they can still be relevant for later analyses.")

# Apply NO filter
before = len(df)

# Important: Int64 allows missing values, int does not
df['pickup_community_area'] = df['pickup_community_area'].astype('Int64')
df['dropoff_community_area'] = df['dropoff_community_area'].astype('Int64')

step2_removed = before - len(df)
print(f"\nRemoved: {step2_removed:,} rows → {len(df):,} remaining")

cleaning_log.append((
    "2 - Community area filter",
    step2_removed,
    "No filter: trips without a CA are kept (see pickup_flag/dropoff_flag)"
))
```

```python
# 1 = pickup community area present, 0 = missing
df["pickup_flag"] = np.where(
    df["pickup_community_area"].notna(),
    1,
    0
)

# 1 = dropoff community area present, 0 = missing
df["dropoff_flag"] = np.where(
    df["dropoff_community_area"].notna(),
    1,
    0
)

# Sanity check
print(df[["pickup_flag", "dropoff_flag"]].value_counts())
```

### Step 2b: Consistent Community Area Centroids

**Problem:** The original coordinate columns mix two geographic resolution levels (census tract vs. community area). This makes direct comparisons across rows misleading.

**Solution:** We derive a CA centroid mapping directly from the data and create new coordinate columns consistently referenced at the CA level.

| Column | Contains | When to use |
|--------|---------|-------------|
| `pickup/dropoff_centroid_lat/lon` | Census tract center **or** community area (CA) center (depending on data availability) | Highest available precision — but note: resolution varies |
| `pickup/dropoff_ca_centroid_lat/lon` | Always the community area center (from this mapping) | Full-coverage, consistent analyses at CA level (heatmaps, aggregations) |


```python
# ── Step 2b: build community area centroid mapping ───────────────────────
#
# BACKGROUND (important for anyone building on this!):
# ───────────────────────────────────────────────
# The pickup/dropoff_centroid_lat/lon columns do NOT always contain the same
# geographic level:
#
#   census_tract present  →  coordinates = census tract centroid
#                             (~0.1 km² resolution, high precision)
#
#   census_tract missing  →  coordinates = community area centroid
#                             (~5-20 km² resolution, coarser)
#
# Direct comparisons of the original coordinates across rows therefore mix two
# precision levels. For analyses at a uniform geographic level (e.g. heatmaps,
# CA aggregations), that's misleading.
#
# SOLUTION:
# ─────────
# Rows WITHOUT a census tract already contain Chicago's official CA centroid.
# We extract these coordinates, build a mapping from them (one row per
# community area), and join it back.
#
# Result: two new consistent columns:
#   pickup_ca_centroid_lat/lon   → always community-area level
#   dropoff_ca_centroid_lat/lon  → always community-area level
#
# WHICH COLUMN TO USE WHEN?
# ───────────────────
#   *_centroid_lat/lon       → original, highest available precision (mixed)
#   *_ca_centroid_lat/lon    → consistent CA level (for full-coverage analyses)

EXPECTED_N_CA = 77  # number of official Chicago community areas

def build_ca_centroid_mapping(df, ca_col, ct_col, lat_col, lon_col):
    # Only rows without a census tract: there, the coordinates = CA centroid
    subset = df[df[ct_col].isna() & df[lat_col].notna() & df[ca_col].notna()]
    median_coords = subset.groupby(ca_col)[[lat_col, lon_col]].median()
    std_coords    = subset.groupby(ca_col)[[lat_col, lon_col]].std()

    # Fallback for CAs that NEVER occur without a census tract (subset has no
    # rows for them): without a fallback, pickup/dropoff_ca_centroid_lat/lon
    # would stay NaN there. We then take the median over ALL rows of that CA
    # (less exact, since some are tract centers instead of the CA center, but
    # better than a gap).
    all_cas     = set(df[ca_col].dropna().unique())
    missing_cas = all_cas - set(median_coords.index)
    if missing_cas:
        fallback_subset = df[df[ca_col].isin(missing_cas) & df[lat_col].notna()]
        fallback_median = fallback_subset.groupby(ca_col)[[lat_col, lon_col]].median()
        median_coords = pd.concat([median_coords, fallback_median]).sort_index()
        print(f"  Note: {len(missing_cas)} CA(s) never had a row without a census tract – "
              f"falling back to the median of all rows for this CA: {sorted(missing_cas)}")

    return median_coords, std_coords

pickup_median,  pickup_std  = build_ca_centroid_mapping(
    df, 'pickup_community_area',  'pickup_census_tract',
    'pickup_centroid_latitude',   'pickup_centroid_longitude')

dropoff_median, dropoff_std = build_ca_centroid_mapping(
    df, 'dropoff_community_area', 'dropoff_census_tract',
    'dropoff_centroid_latitude',  'dropoff_centroid_longitude')

# ── Consistency check: std per CA should be close to 0 ───────────────────
# (same CA = always exactly the same coordinates → std ≈ 0)
# Important: check both pickup AND dropoff std, not just pickup –
# otherwise inconsistent dropoff centroids would go unnoticed.
max_std_lat = max(pickup_std['pickup_centroid_latitude'].max(),
                   dropoff_std['dropoff_centroid_latitude'].max())
max_std_lon = max(pickup_std['pickup_centroid_longitude'].max(),
                   dropoff_std['dropoff_centroid_longitude'].max())
ok = max_std_lat < 0.001 and max_std_lon < 0.001
print(f"Consistency check – coordinate std per community area (pickup & dropoff):")
print(f"  Max lat std:  {max_std_lat:.8f}")
print(f"  Max lon std:  {max_std_lon:.8f}")
print(f"  → {'Consistent – all CAs have stable centroid coordinates' if ok else 'WARNING: spread > 0.001 - please check!'}")

print(f"\nMapping covers: {len(pickup_median)} community areas (pickup), {len(dropoff_median)} (dropoff)")
if len(pickup_median) < EXPECTED_N_CA or len(dropoff_median) < EXPECTED_N_CA:
    print(f"  WARNING: fewer than the expected {EXPECTED_N_CA} Chicago community areas covered!")

# ── Create new columns ──────────────────────────────────────────────────────────────────────────────────
df = df.merge(
    pickup_median.rename(columns={
        'pickup_centroid_latitude':  'pickup_ca_centroid_lat',
        'pickup_centroid_longitude': 'pickup_ca_centroid_lon',
    }),
    left_on='pickup_community_area', right_index=True, how='left',
)
df = df.merge(
    dropoff_median.rename(columns={
        'dropoff_centroid_latitude':  'dropoff_ca_centroid_lat',
        'dropoff_centroid_longitude': 'dropoff_ca_centroid_lon',
    }),
    left_on='dropoff_community_area', right_index=True, how='left',
)

# ── Sample: original vs. CA centroid difference ───────────────────────
print("\nSample of community area 8 (Near North Side) – first 5 trips:")
cols_demo = ['pickup_census_tract', 'pickup_centroid_latitude', 'pickup_ca_centroid_lat']
print(df[df['pickup_community_area'] == 8][cols_demo].head(5).to_string(index=False))
print("\n→ Where census_tract is present: centroid_lat = tract center (can differ)")
print("  Where census_tract is missing: centroid_lat = ca_centroid_lat (identical)")
print(f"\nTotal columns: {df.shape[1]}")

cleaning_log.append((
    "2b - CA centroid mapping",
    0,
    f"{len(pickup_median)} pickup / {len(dropoff_median)} dropoff CAs covered (of expected {EXPECTED_N_CA}); no rows removed"
))
```

### Step 3: Missing Trip Values

Trips without `fare`, `trip_total`, `trip_seconds`, or `trip_miles` are either aborted or recorded erroneously – without these core values, no meaningful analysis is possible. How many trips are affected?

### Step 3c: Tips, Tolls, Extras – Negative Values & Distribution

`tips`, `tolls`, and `extras` can legitimately be 0 — cash payments are recorded without a tip, and many trips have no tolls or surcharges. Negative values, on the other hand, aren't plausible and are removed.


```python
tte_cols = ['tips', 'tolls', 'extras']

print(f"{'Column':<10} {'Null':>8} {'Negative':>10} {'= 0':>10} {'> 0':>10} {'Median':>8} {'Max':>8}")
print("-" * 66)
for col in tte_cols:
    n_null = df[col].isna().sum()
    n_neg  = (df[col] < 0).sum()
    n_zero = (df[col] == 0).sum()
    n_pos  = (df[col] > 0).sum()
    med    = df[col].median()
    mx     = df[col].max()
    print(f"{col:<10} {n_null:>8,} {n_neg:>10,} {n_zero:>10,} {n_pos:>10,} {med:>8.2f} {mx:>8.2f}")

print(f"\n→ tips = 0 is normal: cash payments are recorded without a tip.")
print(f"  tolls = 0 and extras = 0 are also common – only toll roads/surcharges apply to some trips.")

# Distribution of the positive values
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle("Distribution of tips / tolls / extras (values > 0 only, log-scaled)", fontsize=12, fontweight='bold')
for ax, col in zip(axes, tte_cols):
    pos_data = df[col][df[col] > 0].dropna()
    if len(pos_data) == 0:
        ax.set_visible(False)
        continue
    p999 = pos_data.quantile(0.999)
    ax.hist(pos_data.clip(upper=p999 * 1.5), bins=60, color='steelblue', edgecolor='none', alpha=0.85)
    ax.set_title(f"{col}  (n={len(pos_data):,} = {len(pos_data)/len(df)*100:.1f}%)")
    ax.set_xlabel(f'{col} ($)')
    ax.set_ylabel('Frequency')
    ax.set_yscale('log')
plt.tight_layout()
plt.show()

# Remove negative values
before = len(df)
df = df[(df['tips'] >= 0) & (df['tolls'] >= 0) & (df['extras'] >= 0)].copy()
step3c_removed = before - len(df)
print(f"\nNegative tips/tolls/extras removed: {step3c_removed:,} rows → {len(df):,} remaining")

cleaning_log.append((
    "3c - Negative tips/tolls/extras",
    step3c_removed,
    "Rows with negative values removed"
))
```

### Step 3d: Categorical Columns – payment_type & company

```python
# payment_type
print("=== payment_type ===")
n_null_pt = df['payment_type'].isna().sum()
print(f"Missing: {n_null_pt:,}  ({n_null_pt/len(df)*100:.2f}%)\n")
print(df['payment_type'].value_counts(dropna=False).to_string())

# company
print("\n=== company (Top 20) ===")
n_null_co = df['company'].isna().sum()
print(f"Missing: {n_null_co:,}  ({n_null_co/len(df)*100:.2f}%)\n")
print(df['company'].value_counts(dropna=False).head(20).to_string())

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

pt_plot = df['payment_type'].value_counts(dropna=False).head(10)
axes[0].barh(pt_plot.index.astype(str), pt_plot.values, color='steelblue')
axes[0].set_title('payment_type – frequency', fontweight='bold')
axes[0].set_xlabel('Number of trips')
axes[0].invert_yaxis()

co_plot = df['company'].value_counts(dropna=False).head(15)
axes[1].barh(co_plot.index.astype(str), co_plot.values, color='steelblue')
axes[1].set_title('company – top 15', fontweight='bold')
axes[1].set_xlabel('Number of trips')
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print("\n→ No filtering – missing values in payment_type/company are kept as NaN.")

cleaning_log.append((
    "3d - payment_type/company",
    0,
    "No filtering, NaN is kept"
))
```

### Step 4: Outliers – Visualization & Capping

Negative or zero values are physically impossible (no trip with 0 miles or 0 seconds). Very high values *can* be genuine long-distance trips, but are often data errors. Before capping, we look at the distributions and percentiles.


```python
# Chicago taxi base fare (flag drop): $3.25 – any fare/trip total below this is invalid
MIN_FARE = 3.25
MIN_SECONDS = 60

fig, axes = plt.subplots(2, 2, figsize=(14, 8))
fig.suptitle("Distributions before outlier capping (frequency, log-scaled)", fontsize=13, fontweight='bold')

metrics = [
    ('trip_miles',   'Distance (miles)',   axes[0, 0]),
    ('trip_seconds', 'Duration (seconds)', axes[0, 1]),
    ('fare',         'Fare ($)',           axes[1, 0]),
    ('trip_total',   'Trip Total ($)',     axes[1, 1]),
]

print(f"{'Column':<20} {'Min':>8} {'P50':>8} {'P95':>8} {'P99':>8} {'P99.9':>8} {'Max':>10}")
print("-" * 70)

for col, label, ax in metrics:
    data = df[col][df[col] > 0]
    p95, p99, p999 = data.quantile([0.95, 0.99, 0.999])

    print(f"{col:<20} {data.min():>8.1f} {data.median():>8.1f} {p95:>8.1f} {p99:>8.1f} {p999:>8.1f} {data.max():>10.1f}")

    clip_max = p999 * 2
    ax.hist(data.clip(upper=clip_max), bins=80, color='steelblue', edgecolor='none', alpha=0.85)
    ax.axvline(p99,  color='orange', linewidth=1.8, linestyle='--', label=f'P99  = {p99:.1f}')
    ax.axvline(p999, color='red',    linewidth=1.8, linestyle='--', label=f'P99.9 = {p999:.1f}')

    if col == 'trip_seconds':
        ax.axvline(MIN_SECONDS, color='purple', linewidth=1.8, linestyle=':', label=f'Min = {MIN_SECONDS} s')
    elif col in ('fare', 'trip_total'):
        ax.axvline(MIN_FARE, color='purple', linewidth=1.8, linestyle=':', label=f'Min = ${MIN_FARE}')

    ax.set_xlabel(label)
    ax.set_ylabel('Frequency')
    ax.set_yscale('log')
    ax.legend(fontsize=9)
    ax.set_title(col)

plt.tight_layout()
plt.show()

# Lower-bound checks
under_60s   = (df['trip_seconds'] < MIN_SECONDS).sum()
under_fare  = (df['fare']         < MIN_FARE).sum()
under_total = (df['trip_total']   < MIN_FARE).sum()

print(f"\nTrips under {MIN_SECONDS}s duration:                {under_60s:,}  ({under_60s/len(df)*100:.2f}%)")
print(f"  → Too short for a real taxi trip (getting in/out, paying).")
print(f"\nTrips with fare < ${MIN_FARE} (below base fare):    {under_fare:,}  ({under_fare/len(df)*100:.2f}%)")
print(f"Trips with trip total < ${MIN_FARE}:                 {under_total:,}  ({under_total/len(df)*100:.2f}%)")
print(f"  → Chicago's flag drop (base fare on entry) is $3.25.")
print(f"    Anything below that is an aborted, test, or erroneous entry.")
print(f"\n→ Lower bounds: trip_seconds ≥ {MIN_SECONDS}, fare ≥ ${MIN_FARE}, trip_total ≥ ${MIN_FARE}.")
print(f"  Upper bounds: trip_miles and trip_seconds at the 99.9th percentile.")
```

```python
before = len(df)

# Lower bounds: physically impossible or below Chicago's minimum standards
df = df[
    (df['trip_miles']   >  0)          &
    (df['trip_seconds'] >= MIN_SECONDS) &
    (df['fare']         >= MIN_FARE)    &
    (df['trip_total']   >= MIN_FARE)
].copy()
after_lower = len(df)
step4_lower_removed = before - after_lower
print(f"Lower-bound filter removed: {step4_lower_removed:,} rows")
print(f"  (trip_miles > 0, trip_seconds ≥ {MIN_SECONDS}s, fare ≥ ${MIN_FARE}, trip_total ≥ ${MIN_FARE})")

# Upper bounds: 99.9th percentile
cap_miles   = df['trip_miles'].quantile(0.999)
cap_seconds = df['trip_seconds'].quantile(0.999)
df = df[(df['trip_miles'] <= cap_miles) & (df['trip_seconds'] <= cap_seconds)].copy()

removed_upper = after_lower - len(df)
print(f"\nOutlier filter removed:      {removed_upper:,} rows")
print(f"  Distance cap: {cap_miles:.1f} miles")
print(f"  Duration cap: {cap_seconds/60:.0f} minutes")
print(f"\nTotal removed: {before - len(df):,} ({(before - len(df))/before*100:.1f}%)")
print(f"Remaining:     {len(df):,} rows")

cleaning_log.append((
    "4 - Lower bounds",
    step4_lower_removed,
    f"trip_miles > 0, trip_seconds ≥ {MIN_SECONDS}s, fare/trip_total ≥ ${MIN_FARE}"
))
cleaning_log.append((
    "4 - Upper bounds (P99.9)",
    removed_upper,
    f"Distance > {cap_miles:.1f} mi or duration > {cap_seconds/60:.0f} min capped"
))
```
